# Task-Aware Compression: Exploration

This notebook provides a quick, interactive way to view the results generated by the full training and evaluation pipeline.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Set base paths
BASE_DIR = os.path.dirname(os.getcwd())
METRICS_DIR = os.path.join(BASE_DIR, 'results', 'metrics')
PLOTS_DIR = os.path.join(BASE_DIR, 'results', 'plots')
```

### 1. Task-Selector Table
This determines the absolute best method/mode/dimension combination for each downstream task.

```python
selector_path = os.path.join(METRICS_DIR, 'task_selector_table.csv')

if os.path.exists(selector_path):
    df_selector = pd.read_csv(selector_path)
    display(df_selector)
else:
    print('task_selector_table.csv not found. Please run scripts/evaluate.py --task_selector')
```

### 2. Full Results Matrix
A condensed look at the performance of all baseline, task-aware, and task-agnostic runs.

```python
full_results_path = os.path.join(METRICS_DIR, 'full_results_table.csv')

if os.path.exists(full_results_path):
    df_full = pd.read_csv(full_results_path)
    # Sort by STS performance as an example view
    display(df_full.sort_values(by='STS_spearman', ascending=False).head(15))
else:
    print('full_results_table.csv not found. Please run scripts/evaluate.py --all')
```
